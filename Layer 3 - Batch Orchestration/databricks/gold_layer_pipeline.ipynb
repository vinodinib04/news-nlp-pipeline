```python
from pyspark.sql.functions import (
    col,
    explode,
    sha2,
    to_date
)

silver_articles = (
    silver_df
    .select(
        explode(col("articles")).alias("article")
    )
    .select("article.*")
)

silver_with_category = (
    silver_articles
    .withColumn("url_hash", sha2(col("url"), 256))
    .join(
        category_lookup,
        on="url_hash",
        how="left"
    )
    .withColumn(
        "article_date",
        to_date(col("publishedAt"))
    )
)

display(
    silver_with_category
    .select(
        "url_hash",
        "category",
        "article_date",
        "sentiment"
    )
    .limit(20)
)
```

```python
from pyspark.sql.functions import (
    explode,
    date_trunc,
    count
)

top_entities = (
    silver_with_category
    .select(
        "category",
        "publishedAt",
        explode("entities").alias("entity")
    )
    .withColumn(
        "week",
        date_trunc("week", col("publishedAt"))
    )
    .groupBy(
        "week",
        "category",
        "entity"
    )
    .agg(
        count("*").alias("entity_count")
    )
    .orderBy(
        "week",
        "category",
        col("entity_count").desc()
    )
)

display(top_entities)
```

```python
from pyspark.sql.functions import (
    col,
    explode,
    sha2,
    regexp_extract,
    to_date
)

# Load raw data
raw_df = (
    spark.read
    .option("multiLine", "true")
    .json("abfss://news@newsstorage1.dfs.core.windows.net/raw/*.json")
)

# Extract articles and source file path
raw_articles = (
    raw_df
    .select(
        explode(col("articles")).alias("article"),
        col("_metadata.file_path").alias("source_file")
    )
    .select(
        col("article.*"),
        col("source_file")
    )
)

# Build URL hash and category lookup
raw_with_metadata = (
    raw_articles
    .withColumn("url_hash", sha2(col("url"), 256))
    .withColumn(
        "category",
        regexp_extract(
            col("source_file"),
            r"/(tech|business|science|health)-news",
            1
        )
    )
)

category_lookup = (
    raw_with_metadata
    .select("url_hash", "category")
    .dropDuplicates(["url_hash"])
)

# Load Silver data
silver_path = "abfss://news@newsstorage1.dfs.core.windows.net/silver/*.json"

silver_df = (
    spark.read
    .option("multiLine", "true")
    .json(silver_path)
)

# Silver files contain an articles array
silver_articles = (
    silver_df
    .select(
        explode(col("articles")).alias("article")
    )
    .select("article.*")
)

# Join Silver articles to their category
silver_with_category = (
    silver_articles
    .withColumn("url_hash", sha2(col("url"), 256))
    .join(
        category_lookup,
        on="url_hash",
        how="left"
    )
    .withColumn(
        "article_date",
        to_date(col("publishedAt"))
    )
)

print("silver_with_category created successfully")
```

```python
from pyspark.sql.functions import count

sentiment_trends = (
    silver_with_category
    .groupBy(
        "article_date",
        "category",
        "sentiment"
    )
    .agg(
        count("*").alias("article_count")
    )
    .orderBy(
        "article_date",
        "category",
        "sentiment"
    )
)

display(sentiment_trends)
```

```python
from pyspark.sql.functions import (
    explode,
    col,
    count,
    date_sub,
    lit
)

latest_date = (
    silver_with_category
    .select("article_date")
    .agg({"article_date": "max"})
    .collect()[0][0]
)

trending_keywords = (
    silver_with_category
    .filter(
        col("article_date") >= date_sub(
            lit(latest_date),
            6
        )
    )
    .select(
        "category",
        explode("keyPhrases").alias("keyword")
    )
    .groupBy(
        "category",
        "keyword"
    )
    .agg(
        count("*").alias("keyword_count")
    )
    .orderBy(
        "category",
        col("keyword_count").desc()
    )
)

display(trending_keywords)
```

```python
sentiment_trends.write \
    .format("delta") \
    .mode("overwrite") \
    .save("abfss://news@newsstorage1.dfs.core.windows.net/gold/sentiment_trends")

top_entities.write \
    .format("delta") \
    .mode("overwrite") \
    .save("abfss://news@newsstorage1.dfs.core.windows.net/gold/top_entities")

trending_keywords.write \
    .format("delta") \
    .mode("overwrite") \
    .save("abfss://news@newsstorage1.dfs.core.windows.net/gold/trending_keywords")

print("All 3 Gold Delta tables written successfully.")
```

```python
from delta.tables import DeltaTable

def merge_to_delta(df, target_path, merge_condition):
    if DeltaTable.isDeltaTable(spark, target_path):
        target = DeltaTable.forPath(spark, target_path)

        (
            target.alias("target")
            .merge(
                df.alias("source"),
                merge_condition
            )
            .whenMatchedUpdateAll()
            .whenNotMatchedInsertAll()
            .execute()
        )
    else:
        (
            df.write
            .format("delta")
            .mode("overwrite")
            .save(target_path)
        )

    print(f"MERGE completed: {target_path}")
```

```python
# Gold table paths
sentiment_path = "abfss://news@newsstorage1.dfs.core.windows.net/gold/sentiment_trends"
entities_path = "abfss://news@newsstorage1.dfs.core.windows.net/gold/top_entities"
keywords_path = "abfss://news@newsstorage1.dfs.core.windows.net/gold/trending_keywords"

# 1. Sentiment trends
merge_to_delta(
    sentiment_trends,
    sentiment_path,
    """
    target.article_date = source.article_date
    AND target.category = source.category
    AND target.sentiment = source.sentiment
    """
)

# 2. Top entities
merge_to_delta(
    top_entities,
    entities_path,
    """
    target.week = source.week
    AND target.category = source.category
    AND target.entity = source.entity
    """
)

# 3. Trending keywords
merge_to_delta(
    trending_keywords,
    keywords_path,
    """
    target.category = source.category
    AND target.keyword = source.keyword
    """
)

print("All 3 Gold tables merged successfully.")
```
